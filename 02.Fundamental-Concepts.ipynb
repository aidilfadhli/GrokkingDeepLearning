## 2. Theoretical Deep-Dive

### 2.1 The Major Machine Learning Paradigms

Machine learning algorithms are categorized by the nature of the learning feedback:

1. **Supervised Learning**:
   - Learns a mapping function $f: \mathcal{X} \to \mathcal{Y}$ from labeled input-output pairs $\{(x_1, y_1), \dots, (x_n, y_n)\}$.
   - Primary tasks: **Regression** (continuous output) and **Classification** (discrete categorical output).
   - Examples: Medical diagnosis, speech transcription, object detection.

2. **Unsupervised Learning**:
   - Discovers underlying structure or probability density $P(X)$ from unlabeled data $\{x_1, \dots, x_n\}$.
   - Primary tasks: **Clustering**, **Dimensionality Reduction**, **Generative Modeling**.
   - Examples: Customer segmentation (k-Means), PCA, autoencoders.

3. **Parametric vs. Non-Parametric Models**:
   - **Parametric Models**: Summarize data with a fixed number of parameters $\boldsymbol{\theta} = \{w_1, \dots, w_k\}$ independent of dataset size (e.g., Artificial Neural Networks, Linear Regression). Memory footprint is constant at inference time.
   - **Non-Parametric Models**: The number of parameters grows with data size; the algorithm retains all or part of the training set (e.g., $k$-Nearest Neighbors, Decision Trees, Kernel SVMs).

### 2.2 The Universal Deep Learning Paradigm: Predict, Compare, Learn

Every neural network architecture—from single-layer perceptrons to 100-billion-parameter transformers—follows the exact three-stage cycle:

```
      +-------------------------------------------------------------+
      |                     PREDICT                                 |
      |  Given inputs X and current weights W, calculate output:    |
      |                     y_pred = f(X, W)                        |
      +------------------------------+------------------------------+
                                     | 
                                     v
      +-------------------------------------------------------------+
      |                     COMPARE                                 |
      |  Evaluate prediction against ground truth target y_true:    |
      |                 Error = Loss(y_pred, y_true)                |
      +------------------------------+------------------------------+
                                     | 
                                     v
      +-------------------------------------------------------------+
      |                      LEARN                                  |
      |  Calculate gradients and adjust weights to reduce error:    |
      |             W_new = W_old - alpha * (dError/dW)             |
      +-------------------------------------------------------------+
```

- **Predict**: Forward propagation of information.
- **Compare**: Quantifying error via an objective/loss function (e.g., Mean Squared Error).
- **Learn**: Credit assignment and parameter updates via optimization (e.g., Gradient Descent and Backpropagation).

### 2.3 Mathematical Bedrock: Vectors, Matrices, and the Dot Product

Neural networks represent concepts as vectors—ordered collections of numbers in Euclidean space $\mathbb{R}^n$.

#### The Dot Product (Scalar Product)
For two vectors $\mathbf{a}, \mathbf{b} \in \mathbb{R}^n$:

$$\mathbf{a} \cdot \mathbf{b} = \sum_{i=1}^n a_i b_i = a_1 b_1 + a_2 b_2 + \dots + a_n b_n$$

Geometrically, the dot product connects vector lengths and the angle $\theta$ between them:

$$\mathbf{a} \cdot \mathbf{b} = \|\mathbf{a}\| \|\mathbf{b}\| \cos(\theta)$$

**Why this is crucial in Neural Networks:**
1. **Weighted Sum**: If $\mathbf{a}$ is an input vector and $\mathbf{b}$ is a weight vector, $\mathbf{a} \cdot \mathbf{b}$ gives the exact weighted activation of the neuron.
2. **Similarity Metric**: If the weights match the direction of the input vector ($\|\theta\| \to 0^\circ$), the dot product produces its maximum positive output. If orthogonal ($90^\circ$), it outputs zero (complete independence). If opposite ($180^\circ$), it produces maximum negative activation.

## 3. Code Reproduction: Linear Algebra from Scratch vs. NumPy

We now reproduce the core linear algebra operations in pure Python (loops and lists) to demonstrate inner workings, followed by equivalent vectorized NumPy calls.

````python
# 3.1 Vector Elementwise Addition and Multiplication in Pure Python

def elementwise_addition(v1, v2):
    assert len(v1) == len(v2), "Vectors must have identical dimensions!"
    return [v1[i] + v2[i] for i in range(len(v1))]

def elementwise_multiplication(v1, v2):
    assert len(v1) == len(v2), "Vectors must have identical dimensions!"
    return [v1[i] * v2[i] for i in range(len(v1))]

def scalar_multiply(scalar, v):
    return [scalar * x for x in v]

a = [1.0, 2.5, 3.0]
b = [4.0, 0.5, 2.0]

print("Vector a:           ", a)
print("Vector b:           ", b)
print("Elementwise Add:    ", elementwise_addition(a, b))
print("Elementwise Mul:    ", elementwise_multiplication(a, b))
print("Scalar Mult (2.0*a):", scalar_multiply(2.0, a))
````

````python
# 3.2 Vector Dot Product from Scratch vs. NumPy

import numpy as np

def dot_product_scratch(v1, v2):
    assert len(v1) == len(v2), "Vectors must have identical dimensions!"
    total = 0.0
    for i in range(len(v1)):
        total += v1[i] * v2[i]
    return total

# Sample inputs and weights
inputs = [8.5, 0.65, 1.2]
weights = [0.1, 0.2, -0.15]

scratch_result = dot_product_scratch(inputs, weights)
numpy_result = np.dot(np.array(inputs), np.array(weights))

print(f"Dot Product (Scratch): {scratch_result:.6f}")
print(f"Dot Product (NumPy):   {numpy_result:.6f}")
assert np.isclose(scratch_result, numpy_result), "Scratch and NumPy results must match!"
````

````python
# 3.3 Matrix-Vector Multiplication from Scratch vs. NumPy

# In neural networks, a weight matrix maps an input vector to multiple output neurons
# Shape: (num_outputs, num_inputs) * (num_inputs,) -> (num_outputs,)

def matrix_vector_mult_scratch(matrix, vector):
    assert len(matrix[0]) == len(vector), "Matrix column count must match vector length!"
    output = []
    for row in matrix:
        output.append(dot_product_scratch(row, vector))
    return output

# Weights matrix: 3 output neurons, each connected to 3 inputs
weight_matrix = [
    [0.1, 0.2, -0.1],  # weights for output 1 (e.g., hurt?)
    [-0.3, 0.5, 0.1],  # weights for output 2 (e.g., win?)
    [0.0, 0.1, 0.2]    # weights for output 3 (e.g., sad?)
]

inputs = [8.5, 0.65, 1.2]

output_scratch = matrix_vector_mult_scratch(weight_matrix, inputs)
output_numpy = np.dot(np.array(weight_matrix), np.array(inputs))

print("Matrix-Vector Multiplications:")
print("Scratch Output:", [round(x, 4) for x in output_scratch])
print("NumPy Output:  ", np.round(output_numpy, 4))
assert np.allclose(output_scratch, output_numpy)
````

````python
# 3.4 Matrix-Matrix Multiplication (Batching) and Vectorization Speedup
import time

# Benchmark comparing pure Python nested loops vs. NumPy BLAS acceleration
dim = 200
A = np.random.randn(dim, dim)
B = np.random.randn(dim, dim)
A_list = A.tolist()
B_list = B.tolist()

# Pure Python matrix multiplication (O(N^3))
t0 = time.time()
C_scratch = [[0.0 for _ in range(dim)] for _ in range(dim)]
for i in range(dim):
    for j in range(dim):
        for k in range(dim):
            C_scratch[i][j] += A_list[i][k] * B_list[k][j]
t_scratch = time.time() - t0

# NumPy optimized GEMM (using C / BLAS / AVX instructions)
t0 = time.time()
C_numpy = np.dot(A, B)
t_numpy = time.time() - t0

print(f"Dimension {dim}x{dim}:")
print(f"Pure Python Time: {t_scratch:.4f} seconds")
print(f"NumPy Time:       {t_numpy:.6f} seconds")
print(f"Speedup Factor:   {t_scratch / t_numpy:.1f}x faster with NumPy!")
````
