## 2. Theoretical Deep-Dive

### 2.1 Gradient Descent with Multiple Inputs

Consider a neuron accepting $m$ inputs $\mathbf{x} = [x_1, \dots, x_m]^T$ with weights $\mathbf{w} = [w_1, \dots, w_m]^T$:

$$\hat{y} = \sum_{i=1}^m x_i w_i, \quad \text{Error} = (\hat{y} - y)^2$$

To compute the adjustment for each weight $w_i$, we calculate the partial derivative with respect to that specific weight, treating all other weights as constants:

$$\frac{\partial \text{Error}}{\partial w_i} = \frac{\partial \text{Error}}{\partial \hat{y}} \cdot \frac{\partial \hat{y}}{\partial w_i} = 2(\hat{y} - y) \cdot x_i$$

$$\text{delta} = \hat{y} - y = \text{pred} - \text{actual}$$
$$\text{weight\_delta}_i = \text{delta} \cdot x_i$$
$$w_i \leftarrow w_i - \alpha \cdot \text{weight\_delta}_i$$

**A Fundamental Insight:**  
Notice that **every weight shares the exact same `delta`!** The shared `delta` reflects the overall error of the network's prediction. What differentiates the update of each weight is the individual input feature $x_i$ connected to that weight.

### 2.2 The Geometry of Error Surfaces & 'Frozen Weights'

When optimizing multiple parameters:
- With 1 weight, the error surface is a 2D parabola: $\text{Error} = f(w)$.
- With 2 weights, the error surface is a 3D bowl: $\text{Error} = f(w_1, w_2)$.
- With $M$ weights, the error surface is an $(M+1)$-dimensional paraboloid.

#### The 'Frozen Weights' Thought Experiment
What happens if we freeze $w_2$ and $w_3$ and only update $w_1$?
- $w_1$ will adjust until it reaches the lowest point along its single 1D slice of the bowl.
- However, that point is **not** the true global minimum of the entire system!
- When all weights update concurrently, each weight adjusts in coordination, moving down the steepest gradient direction: $\nabla \text{Error} = [\frac{\partial E}{\partial w_1}, \dots, \frac{\partial E}{\partial w_m}]^T$.

### 2.3 Multiple Inputs and Multiple Outputs: The Outer Product

When extending to $m$ inputs and $k$ outputs, the weight matrix $\mathbf{W} \in \mathbb{R}^{k \times m}$ maps $\mathbf{x} \to \hat{\mathbf{y}}$:

$$\hat{\mathbf{y}} = \mathbf{W} \mathbf{x}$$

For each output neuron $j \in \{1, \dots, k\}$:
$$\text{delta}_j = \hat{y}_j - y_j$$

The gradient with respect to weight $w_{ji}$ (connecting input $i$ to output $j$) is:
$$\frac{\partial \text{Error}}{\partial w_{ji}} = \text{delta}_j \cdot x_i$$

In matrix form, the full weight gradient matrix is the **outer product** of the output error vector $\boldsymbol{\delta}$ and the input vector $\mathbf{x}$:

$$\Delta \mathbf{W} = \boldsymbol{\delta} \otimes \mathbf{x} = \boldsymbol{\delta} \mathbf{x}^T = 
\begin{bmatrix} \delta_1 \\ \delta_2 \\ \vdots \\ \delta_k \end{bmatrix}
\begin{bmatrix} x_1 & x_2 & \dots & x_m \end{bmatrix}
= 
\begin{bmatrix} 
\delta_1 x_1 & \delta_1 x_2 & \dots & \delta_1 x_m \\ 
\delta_2 x_1 & \delta_2 x_2 & \dots & \delta_2 x_m \\ 
\vdots & \vdots & \ddots & \vdots \\ 
\delta_k x_1 & \delta_k x_2 & \dots & \delta_k x_m 
\end{bmatrix}$$

$$\mathbf{W} \leftarrow \mathbf{W} - \alpha \cdot \Delta \mathbf{W}$$

## 3. Code Reproduction

We now reproduce the multi-weight gradient descent algorithms from Chapter 5.

```python
# 3.1 Gradient Descent with Multiple Inputs (Pure Python Primitives)

def w_sum(a, b):
    assert len(a) == len(b)
    output = 0.0
    for i in range(len(a)):
        output += (a[i] * b[i])
    return output

def ele_mul(number, vector):
    output = [0.0] * len(vector)
    for i in range(len(vector)):
        output[i] = number * vector[i]
    return output

def neural_network(input_vec, weights):
    return w_sum(input_vec, weights)

# Baseball dataset inputs for Game 1
toes =  [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]
win_or_lose_binary = [1, 1, 0, 1]

weights = [0.1, 0.2, -0.1]
input_val = [toes[0], wlrec[0], nfans[0]]
true_val = win_or_lose_binary[0]

pred = neural_network(input_val, weights)
error = (pred - true_val) ** 2
delta = pred - true_val
weight_deltas = ele_mul(delta, input_val)

alpha = 0.01
for i in range(len(weights)):
    weights[i] -= alpha * weight_deltas[i]

print("One-Step Multi-Input Gradient Descent:")
print(f"Prediction:    {pred:.6f}")
print(f"Error:         {error:.6f}")
print(f"Delta:         {delta:.6f}")
print(f"Weight Deltas: {[round(x, 6) for x in weight_deltas]}")
print(f"New Weights:   {[round(w, 6) for w in weights]}")
```

```python
# 3.2 Watching Several Steps of Learning with Multiple Inputs

weights = [0.1, 0.2, -0.1]
alpha = 0.01

print("Training Multiple Weights over 25 Iterations:")
for iteration in range(25):
    pred = neural_network(input_val, weights)
    error = (pred - true_val) ** 2
    delta = pred - true_val
    weight_deltas = ele_mul(delta, input_val)
    
    for i in range(len(weights)):
        weights[i] -= alpha * weight_deltas[i]
        
    if iteration % 5 == 0 or iteration == 24:
        print(f"Iter {iteration:2d} | Error: {error:10.8f} | Pred: {pred:8.6f} | Weights: {[round(w, 4) for w in weights]}")
```

```python
# 3.3 The 'Frozen Weights' Experiment: Locking weight[0]

weights = [0.1, 0.2, -0.1]
alpha = 0.3

print("Training with Weight 0 Frozen (weights[0] remains 0.1):")
for iteration in range(25):
    pred = neural_network(input_val, weights)
    error = (pred - true_val) ** 2
    delta = pred - true_val
    weight_deltas = ele_mul(delta, input_val)
    
    # Force weight_deltas[0] to 0 to simulate frozen weight
    weight_deltas[0] = 0.0
    
    for i in range(len(weights)):
        weights[i] -= alpha * weight_deltas[i]
        
    if iteration % 5 == 0 or iteration == 24:
        print(f"Iter {iteration:2d} | Error: {error:10.8f} | Pred: {pred:8.6f} | Weights: {[round(w, 4) for w in weights]}")
```

```python
# 3.4 Gradient Descent with Multiple Outputs

weights_multi_out = [0.3, 0.2, 0.9]  # [hurt, win, sad]
wlrec_inp = 0.65
true_targets = [0.1, 1.0, 0.1]       # [actual_hurt, actual_win, actual_sad]
alpha = 0.1

print("Training Single Input -> Multiple Outputs:")
for iteration in range(15):
    pred = [wlrec_inp * w for w in weights_multi_out]
    error = [(p - t) ** 2 for p, t in zip(pred, true_targets)]
    delta = [p - t for p, t in zip(pred, true_targets)]
    weight_deltas = [d * wlrec_inp for d in delta]
    
    for i in range(len(weights_multi_out)):
        weights_multi_out[i] -= alpha * weight_deltas[i]
        
    if iteration % 3 == 0 or iteration == 14:
        print(f"Iter {iteration:2d} | Total Error: {sum(error):8.6f} | Preds: {[round(p, 4) for p in pred]}")
```

```python
# 3.5 Full Matrix Gradient Descent (Multiple Inputs -> Multiple Outputs)
import numpy as np

# Weights mapping (toes, wlrec, nfans) -> (hurt, win, sad)
weights = np.array([
    [0.1, 0.1, -0.3],  # hurt?
    [0.1, 0.2, 0.0],   # win?
    [0.0, 1.3, 0.1]    # sad?
])

inputs = np.array([toes[0], wlrec[0], nfans[0]])
targets = np.array([0.0, 1.0, 0.1])
alpha = 0.01

print("NumPy Full Matrix Gradient Descent:")
for iteration in range(20):
    # Forward pass
    pred = weights.dot(inputs)
    error = np.sum((pred - targets) ** 2)
    delta = pred - targets
    
    # Outer product calculation: delta (3x1) * inputs (1x3) -> (3x3)
    weight_deltas = np.outer(delta, inputs)
    
    # Parameter update
    weights -= alpha * weight_deltas
    
    if iteration % 4 == 0 or iteration == 19:
        print(f"Iter {iteration:2d} | Total Error: {error:10.8f} | Predictions: {np.round(pred, 4)}")

print("\nFinal Converged Predictions:", np.round(weights.dot(inputs), 4))
print("Ground Truth Targets:       ", targets)
```

## 4. Chapter 5 Summary & Key Takeaways

| Concept | Mathematical Equation | Key Takeaway |
|:---|:---:|:---|
| **Multi-Input Gradients** | $\frac{\partial E}{\partial w_i} = \text{delta} \cdot x_i$ | All weights share the same error delta, but their gradients scale proportionally to their individual input signal. |
| **Error Surface** | $E(\mathbf{w}) = (\mathbf{w}^T \mathbf{x} - y)^2$ | In higher dimensions, error forms a bowl/paraboloid where simultaneous updates follow the gradient vector $\nabla E$. |
| **Outer Product Update** | $\Delta \mathbf{W} = \boldsymbol{\delta} \mathbf{x}^T$ | Efficiently evaluates all parameter gradients across multiple inputs and outputs simultaneously. |
